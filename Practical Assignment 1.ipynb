#### 01. Write a Python program to read an image named "flower.jpg" using OpenCV and display it in a window titled "Original Image".

```python
import cv2
import matplotlib.pyplot as plt

# Read the image using OpenCV 
img = cv2.imread('flower.jpg')

if img is None:
    print("Error: Could not load 'flower.jpg'. Make sure the file is in the same folder.")
else:
    # OpenCV loads BGR, Matplotlib expects RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    
    # Display the image inline
    plt.imshow(img_rgb)
    plt.axis('off')  # Hide coordinate axes
    plt.show()
```

#### 05. Write a Python program using OpenCV that resizes an image to 50% of its original width and height.

```python
# Load original image
img_cv = cv2.imread('flower.jpg')

# Obtain height and width from shape tuple
h, w = img_cv.shape[:2]

# Compute target dimensions (50% scale factor)
new_w = int(w * 0.5)
new_h = int(h * 0.5)

# Resize using cv2.resize with dimensions order (width, height)
resized = cv2.resize(img_cv, (new_w, new_h), interpolation=cv2.INTER_LINEAR)

print(f"Original Dimensions: ({w}, {h})")
print(f"Resized Dimensions (50%): ({new_w}, {new_h})")
```

```python
import matplotlib.pyplot as plt

# Convert BGR to RGB so colors display accurately
resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)

# Display the resized image
plt.imshow(resized_rgb)
plt.title(f"Resized ({new_w}x{new_h})")
plt.axis('off')  # Hide coordinate axes
plt.show()
```

#### 06. Write a Python code snippet to crop the central 200×200 pixel area of an image and display it using Matplotlib.

```python
#Crop central 200x200 area and display using Matplotlib
import matplotlib.pyplot as plt

# Load image
img_cv = cv2.imread('flower.jpg')

# Find center point of the image
h, w = img_cv.shape[:2]
center_x, center_y = w // 2, h // 2

# Slice region of interest: [y1:y2, x1:x2]
cropped = img_cv[center_y - 100 : center_y + 100, center_x - 100 : center_x + 100]

# Display cropped ROI using Matplotlib (convert BGR to RGB)
plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
plt.title('Central 200x200 Cropped Area')
plt.axis('off')
plt.show()
```

#### 07. Write a short program that rotates an image by 90 degrees clockwise using OpenCV.

```python
#Rotate image 90 degrees clockwise using OpenCV
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread('flower.jpg')

# Rotate 90 degrees clockwise
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

# Convert BGR to RGB for correct color display in Matplotlib
rotated_rgb = cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)

# Display rotated result inline
plt.imshow(rotated_rgb)
plt.axis('off')
plt.show()
```

#### 08. Write a Python statement to flip the image “car.jpg” left-to-right using OpenCV and display the result.

```python
#Flip "car.jpg" left-to-right (horizontal mirror) using OpenCV
import cv2
import matplotlib.pyplot as plt

# Load image
img_car = cv2.imread('car.jpg')

# Flip horizontally: flipCode 1 corresponds to left-to-right mirror
flip_horizontal = cv2.flip(img_car, 1)

# Display the flipped image
plt.imshow(cv2.cvtColor(flip_horizontal, cv2.COLOR_BGR2RGB))
plt.title('Horizontal Flip')
plt.axis('off')
plt.show()
```

#### 09. Write a Python program that converts an image from RGB to HSV colour space, and display each channel (Hue, Saturation, Value) separately using Matplotlib.

```python
#Convert RGB to HSV and display Hue, Saturation, Value channels separately
import cv2
import matplotlib.pyplot as plt

# Read image
img_cv = cv2.imread('car.jpg')

# Convert BGR colour space to HSV
hsv = cv2.cvtColor(img_cv, cv2.COLOR_BGR2HSV)

# Split into individual channels
h, s, v = cv2.split(hsv)

# Display H, S, and V channels side-by-side using Matplotlib
plt.figure(figsize=(12, 3))
for i, (ch_img, title) in enumerate(zip([h, s, v], ['Hue', 'Saturation', 'Value'])):
    plt.subplot(1, 3, i + 1)
    plt.imshow(ch_img, cmap='gray')
    plt.title(title)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

#### 10. Write a short Python code snippet to compute the Mean Squared Error (MSE) between two grayscale images, "img1.jpg" and "img2.jpg".

```python
# Program 10: Compute MSE between "img1.jpg" and "img2.jpg"
import cv2
from skimage.metrics import mean_squared_error

# Load both images in grayscale mode
img1 = cv2.imread('flower.jpg', cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread('car.jpg', cv2.IMREAD_GRAYSCALE)

# Ensure both images exist and loaded correctly
if img1 is None or img2 is None:
    print("Error: One or both image files could not be loaded.")
else:
    # Resize img2 to match the dimensions of img1 (width, height)
    h1, w1 = img1.shape
    img2_resized = cv2.resize(img2, (w1, h1), interpolation=cv2.INTER_LINEAR)

    # Compute Mean Squared Error
    mse_value = mean_squared_error(img1, img2_resized)

    print(f"Original img1 shape: {img1.shape}")
    print(f"Resized img2 shape: {img2_resized.shape}")
    print(f"Mean Squared Error (MSE): {mse_value:.2f}")
```
